# 01 — Data Exploration
Explore raw GEE-exported raster layers.

```python
import rasterio
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from pathlib import Path

BAND_NAMES = ['Elevation','Slope','Aspect','TWI','NDVI','NDWI','Rainfall','Soil','LST']
RAW_TIF = Path('../data/raw/Pothohar_Groundwater_Features.tif')
```

```python
with rasterio.open(RAW_TIF) as src:
    data = src.read()
    print('Shape:', data.shape)
    print('CRS:', src.crs)
    print('Resolution:', src.res)
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
for i, (ax, name) in enumerate(zip(axes.flat, BAND_NAMES)):
    band = data[i]
    im = ax.imshow(band, cmap='viridis')
    ax.set_title(name, fontsize=12)
    ax.axis('off')
    plt.colorbar(im, ax=ax, shrink=0.8)
plt.suptitle('All Feature Layers — Pothohar Plateau', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('../outputs/figures/all_layers.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
flat = data.reshape(len(BAND_NAMES), -1).T
df = pd.DataFrame(flat, columns=BAND_NAMES).dropna()
print(df.describe())
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.sample(10000).corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```
